```python
import os
import re
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, StratifiedShuffleSplit
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import classification_report, confusion_matrix


# init the lable by filename
code_to_value_special = {
    12: 11,  # hundred
    13: 12,  # 1k
    14: 13,  # 10k
    15: 14   # 100m
}

# find label in images' filename and match with images
image_folder = './Chinese_MINST_Dataset/data/data'
image_paths = []
labels = []
for filename in os.listdir(image_folder):
    if filename.endswith('.jpg'):
        image_paths.append(os.path.join(image_folder, filename))
        match = re.match(r'.*_(\d+)\.jpg', filename)
        if not match:
            raise ValueError(f"{filename} name error")
        code = int(match.group(1))
        if 1 <= code <= 11:
            value = code - 1       # the charactors equals to code-1
        elif code in code_to_value_special:
            value = code_to_value_special[code]  # for special labels 
        else:
            raise ValueError(f"unknown: {code}")
        labels.append(value)

labels = np.array(labels)

# load images and reshape 
def load_images(paths):
    images = []
    for path in paths:
        img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)  # change into gray
        if img is None:
            raise FileNotFoundError(f"no file: {path}")
        img_resized = cv2.resize(img, (64, 64))
        images.append(img_resized.flatten()) # reshape into 1D
    return np.array(images)

X_all = load_images(image_paths)
y_all = labels

# split the images into training sets and testing sets
sss_test = StratifiedShuffleSplit(n_splits=1, test_size=1000, random_state=42)
for train_val_idx, test_idx in sss_test.split(X_all, y_all):
    X_test = X_all[test_idx]
    y_test = y_all[test_idx]
    X_train_val = X_all[train_val_idx]
    y_train_val = y_all[train_val_idx]

def sample_train(X, y, train_size): # make sure that the testing set and training set will not covered
    sss = StratifiedShuffleSplit(n_splits=1, train_size=train_size, random_state=42)
    for train_idx, _ in sss.split(X, y):
        return X[train_idx], y[train_idx]

X_train_5k, y_train_5k = sample_train(X_train_val, y_train_val, 5000)
X_train_10k, y_train_10k = sample_train(X_train_val, y_train_val, 10000)

# draw confusion matrix
def draw_confusion_matrix(cm, name, classes):
    plt.imshow(cm, interpolation='nearest', cmap=plt.cm.Blues)
    plt.title(f'confusion matrix: {name}')
    plt.colorbar()
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)

    thresh = cm.max() / 2.
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, format(cm[i, j], 'd'),
                     ha="center", va="center",
                     color="white" if cm[i, j] > thresh else "black")

    plt.ylabel('real')
    plt.xlabel('pred')
    plt.tight_layout()
    plt.show()

# training and evaluating
def train_and_evaluate(model, X_train, y_train, X_test, y_test, name):

    # use fit() and predict()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    # show classification report
    print(f'{name} report:')
    print(classification_report(y_test, y_pred, digits=4))
    cm = confusion_matrix(y_test, y_pred)
    draw_confusion_matrix(cm, name, classes=range(15))

# init the models
knn = KNeighborsClassifier(n_neighbors=3)
dt = DecisionTreeClassifier(random_state=42)
sgd = SGDClassifier(max_iter=250, random_state=42)

# 5000 training
train_and_evaluate(knn, X_train_5k, y_train_5k, X_test, y_test, "KNN (5k)")
train_and_evaluate(dt, X_train_5k, y_train_5k, X_test, y_test, "DTree (5k)")
train_and_evaluate(sgd, X_train_5k, y_train_5k, X_test, y_test, "SGD (5k)")

# 10000 training
train_and_evaluate(knn, X_train_10k, y_train_10k, X_test, y_test, "KNN (10k)")
train_and_evaluate(dt, X_train_10k, y_train_10k, X_test, y_test, "DTree (10k)")
train_and_evaluate(sgd, X_train_10k, y_train_10k, X_test, y_test, "SGD (10k)")
```

## Conclusion

- **All models improved accuracy with increased training size**
- **KNN** showing the most obvious improvement from ~34.6% to ~42.6%.
- **Decision Tree** maintained stable but the accuracy is low
- **SGD** provided a balanced performance, but the accuracy is lower then KNN.
- KNN is the best model in this task
